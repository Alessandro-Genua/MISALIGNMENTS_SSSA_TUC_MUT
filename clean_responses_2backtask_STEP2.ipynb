This code cleans and sorts the data obtained from the STEP 1 code. It provides a cleaned file with just two coulumn for each subject and one for each conditions. One is for aspcted responses and another one is for given responses.


```python
from pathlib import Path
import pandas as pd

# ============================================================
# PATH
# ============================================================

base_path = Path(r"C:\Users\a.genua\OneDrive - Scuola Superiore Sant'Anna\PROGETTI\Progetti in corso\Articoli in corso\RA-L_&_Sanseverino\FlightSimulator")

input_csv = base_path / "2back_human_factors_export_final.csv"
output_csv = base_path / "2back_responses_only_final.csv"

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(input_csv)

print("Shape originale:", df.shape)

# ============================================================
# FILTRO COLONNE
# ============================================================

# tieni solo colonne expected + subject
cols_keep = [
    col for col in df.columns
    if ("expected_response" in col or "subject_response" in col)
]

df_filtered = df[cols_keep].copy()

print("Shape filtrato:", df_filtered.shape)

# ============================================================
# SALVATAGGIO
# ============================================================

df_filtered.to_csv(output_csv, index=False, encoding="utf-8-sig")

print(f"\nCSV salvato in:\n{output_csv}")
```
